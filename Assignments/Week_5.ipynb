```python
import pandas as pd
```

```python
#b.
aapl = pd.read_csv('aapl.csv')
amzn = pd.read_csv('amzn.csv')
fb = pd.read_csv('fb.csv')
goog = pd.read_csv('goog.csv')
nflx = pd.read_csv('nflx.csv')
```

```python
#c.
aapl['ticker'] = 'AAPL'
amzn['ticker'] = 'AMZN'
fb['ticker'] = 'FB'
goog['ticker'] = 'GOOG'
nflx['ticker'] = 'NFLX'
```

```python
#d.
faanf = pd.concat([aapl, amzn, fb, goog, nflx], ignore_index=True)
faanf.head()
```

```python
#e.
faanf.to_csv('faanf.csv', index=False)
```

```python
#f.
faanf = pd.read_csv('faanf.csv')

#Convert date to datetime
faanf['date'] = pd.to_datetime(faanf['date'])

#Convert volume to integer
faanf['volume'] = faanf['volume'].astype(int)

#Sort by date and ticker
faanf = faanf.sort_values(by=['date', 'ticker']).reset_index(drop=True)

faanf.head(10)
```
