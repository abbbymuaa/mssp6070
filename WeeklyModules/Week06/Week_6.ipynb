```python

import pandas as pd
from google.colab import files



df = pd.read_csv("/content/MSSP6070/data/Student Performance Data.csv")

# Define a lambda function that returns a new DataFrame
transform_students = lambda data: data.assign(
    average_score=data[
        ['math score', 'reading score', 'writing score']
    ].mean(axis=1).round(2),

    score_category=data[
        ['math score', 'reading score', 'writing score']
    ].mean(axis=1).apply(
        lambda score: 'High' if score >= 80
        else ('Medium' if score >= 60 else 'Low')
    )
)

# Apply the lambda function
new_df = transform_students(df)

display(new_df.head())

print("Original DataFrame:", df.shape)
print("New DataFrame:", new_df.shape)

# Export the new DataFrame
new_df.to_csv("student_performance_transformed.csv", index=False)
files.download("student_performance_transformed.csv")
```
