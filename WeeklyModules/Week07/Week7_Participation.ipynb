```python
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files

uploaded = files.upload()

df = pd.read_csv("InternetSales.csv", encoding="cp1252")

#convert OrderDate to datetime
df["OrderDate"] = pd.to_datetime(df["OrderDate"])

#extract the year from OrderDate
df["Year"] = df["OrderDate"].dt.year

#calculate total sales for each year
yearly_sales = df.groupby("Year")["SalesAmount"].sum()

#create a bar chart
plt.figure(figsize=(10, 6))

plt.bar(yearly_sales.index.astype(str), yearly_sales.values,
        color="steelblue")

plt.title("Yearly Internet Sales")
plt.xlabel("Year")
plt.ylabel("Total Sales ($)")

plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.savefig("Yearly_Internet_Sales.pdf", format="pdf",
            bbox_inches="tight")
plt.show()

files.download("Yearly_Internet_Sales.pdf")
```
